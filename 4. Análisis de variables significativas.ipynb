```python
full_transactions = spark.read.format("delta").table("hive_metastore.default.business_full_transactions")
full_transactions.limit(5).display()
```

```python
# Cambiamos la columna IsFraud a 0 y 1 y lo mismo en Geneder.
from pyspark.sql.functions import when

full_transactions = full_transactions.withColumn("IsFraud", when(full_transactions['IsFraud'] == "Yes",1).otherwise(0))
full_transactions = full_transactions.withColumn("Gender", when(full_transactions['Gender'] == "Male",1).otherwise(0))
```

```python
# Eliminamos las columnas Person y ID del usuario
full_transactions = full_transactions.drop("Zipcode","Zip","MerchantName","MerchantCity",'Expires', 'ExpiresMonth', ' AcctOpenMonth',' AcctOpeYear','CardNumber', 'CVV',  'RetirementAge','Address','Apartment','City', 'Latitude','Longitude')
```

```python
from pyspark.sql.types import StringType

schema = full_transactions.schema
string_columns = [field.name for field in schema.fields if field.dataType == StringType()]

print(string_columns)
```

```python
from pyspark.ml.feature import OneHotEncoder, StringIndexer
from pyspark.ml import Pipeline

one_hot_cols = ['UseChip', 'RangoHoras', 'DiaDeSemana', 'CardBrand', 'CardType', 'CardsIssued']

# Aplicamos StringIndexer a las columnas categóricas
indexers = [StringIndexer(inputCol=col, outputCol=f"{col}_index", handleInvalid="keep") for col in one_hot_cols]
```

```python
one_hot_encoders = [OneHotEncoder(inputCol=f"{col}_index", outputCol=f"{col}_encoded") for col in one_hot_cols]

# Creamos y aplicamos el pipeline
pipeline = Pipeline(stages=indexers + one_hot_encoders)
df_transformed = pipeline.fit(full_transactions).transform(full_transactions)
```

```python
df_transformed = df_transformed.drop(
    *one_hot_cols, *[f'{i}_index' for i in one_hot_cols]
)
df_transformed.limit(5).display()
```

```python
one_hot_cols = ['MerchantState', 'State']

# Aplicamos StringIndexer a las columnas categóricas
indexers = [StringIndexer(inputCol=col, outputCol=f"{col}_index", handleInvalid="keep") for col in one_hot_cols]
```

```python
pipeline = Pipeline(stages=indexers )
df_transformed = pipeline.fit(df_transformed).transform(df_transformed)
```

```python
df_transformed = df_transformed.drop("State","MerchantState")
```

```python
schema = df_transformed.schema
string_columns = [field.name for field in schema.fields if field.dataType == StringType()]

print(string_columns)
```

```python
df_transformed = df_transformed.drop("Person")
df_transformed.limit(5).display()
```

```python
from pyspark.ml.feature import PCA, VectorAssembler, StandardScaler
from pyspark.ml.linalg import Vectors

# List of feature columns to be included in the vector
feature_columns = ['User',
 'Card',
 'Amount',
 'MCC',
 'Errors',
 'IsFraud',
 'ExpiresYear',
 'HasChip',
 'CreditLimit',
 'AcctOpenMonth',
 'AcctOpenYear',
 'YearPINlastChanged',
 'CardOnDarkWeb',
 'CurrentAge',
 'BirthYear',
 'BirthMonth',
 'Gender',
 'PerCapitaIncomeZipcode',
 'YearlyIncomePerson',
 'TotalDebt',
 'FICOScore',
 'NumCreditCards',
 'UseChip_encoded',
 'RangoHoras_encoded',
 'DiaDeSemana_encoded',
 'CardBrand_encoded',
 'CardType_encoded',
 'CardsIssued_encoded',
 'MerchantState_index',
 'State_index']

# Assemble the feature columns into a feature vector
assembler = VectorAssembler(inputCols=feature_columns, outputCol="features_vector2", handleInvalid="skip")
df_transformed = assembler.transform(df_transformed)

# Standar scaler
scaler = StandardScaler(inputCol="features_vector2", outputCol="features_vector_scaled", withMean=True, withStd=True)
scaler_model = scaler.fit(df_transformed)
df_scaled = scaler_model.transform(df_transformed)
```

```python
# Configure the PCA model
pca = PCA(k=53, inputCol="features_vector_scaled", outputCol="pca_features")

# Fit the PCA model and transform the data
pca_model = pca.fit(df_scaled)
df_pca = pca_model.transform(df_scaled)

# Display the result
display(df_pca)
```

```python
# Obtener la varianza explicada por cada componente principal
explained_variance = pca_model.explainedVariance.toArray()

# Calcular la varianza explicada acumulada
cumulative_variance = explained_variance.cumsum()

# Mostrar resultados
for i, var in enumerate(cumulative_variance):
    print(f"k={i+1}, Varianza explicada acumulada: {var:.4f}")
```

# Porcentaje de transacciones fraudulentas por año de apertura

```python
year_fraud_count = full_transactions.groupBy('AcctOpenYear', 'IsFraud').count()
year_fraud_total = full_transactions.groupBy('AcctOpenYear').count().withColumnRenamed('count', 'total')
year_fraud_percentage = year_fraud_count.join(year_fraud_total, 'AcctOpenYear') \
    .withColumn('percentage', (F.col('count') / F.col('total')) * 100)
```

```python
year_fraud_percentage_filtered = year_fraud_percentage.filter(F.col('IsFraud') == 1)
display(year_fraud_percentage_filtered.select('AcctOpenYear', 'percentage'))
```

```python
year_fraud_percentage_filtered = year_fraud_percentage.filter(F.col('IsFraud') == 1)
display(year_fraud_percentage_filtered.select('ExpiresYear', 'percentage'))
```
