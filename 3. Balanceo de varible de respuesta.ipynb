```python
sub_transaction = full_transactions.filter(
    (full_transactions.IsFraud == "Yes") |
    (full_transactions.Amount > 0) |
    (full_transactions.Amount < 5)
    )
```

```python
FRACTION = {"Yes": 0.0417, "No": 0.041}
sub_1000_to_1_proportion = full_transactions.sampleBy("IsFraud", FRACTION)
```

```python
class_counts = sub_1000_to_1_proportion.groupBy("IsFraud").count()
# Mostrar el resultado
display(class_counts)
```

```python
sub_1000_to_1_proportion.write.format('delta').mode('overwrite').saveAsTable('sample_1000_to_1_proportion')
```
